```python
import os
import cv2
import albumentations as A

# Path dataset
input_folder = r"D:\Document\Kuliah\Semster 5\AI\TUBES AI\coba_dataset_image"
output_folder = r"D:\Document\Kuliah\Semster 5\AI\TUBES AI\augmented_image"

# Fungsi untuk membuat folder jika belum ada
def create_folder(path):
    if not os.path.exists(path):
        os.makedirs(path)

# Augmentasi gambar
augmentation_pipeline = [
    ("HorFlip", A.HorizontalFlip(p=1.0)),
    ("VerFlip", A.VerticalFlip(p=1.0)),
    ("RandomBrightnessContrast", A.RandomBrightnessContrast(p=1.0)),
    ("GaussBlur", A.GaussianBlur(blur_limit=(3, 7), p=1.0)),
    ("Rotate", A.Rotate(limit=45, p=1.0))
]

# Proses augmentasi
for category in os.listdir(input_folder):
    input_category_path = os.path.join(input_folder, category)
    output_category_path = os.path.join(output_folder, category)
    create_folder(output_category_path)

    for image_name in os.listdir(input_category_path):
        input_image_path = os.path.join(input_category_path, image_name)

        # Load gambar
        image = cv2.imread(input_image_path)
        if image is None:
            continue
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        # Terapkan setiap augmentasi
        for aug_name, aug in augmentation_pipeline:
            augmented = aug(image=image)
            augmented_image = augmented["image"]

            # Simpan dengan nama format baru
            base_name = os.path.splitext(image_name)[0]
            output_image_name = f"{base_name}aug{aug_name}.jpg"
            output_image_path = os.path.join(output_category_path, output_image_name)

            # Konversi ke BGR untuk menyimpan dengan OpenCV
            augmented_image = cv2.cvtColor(augmented_image, cv2.COLOR_RGB2BGR)
            cv2.imwrite(output_image_path, augmented_image)

print("Augmentasi selesai!")
```
