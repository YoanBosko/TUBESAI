```python
import os
import cv2
import albumentations as A
```

```python
# Menetapkan path untuk dataset
image_path = r'D:\Document\Kuliah\Semster5\AI\TUBESAI\dataset_image'  # Path folder yang berisi sub-folder kategori gambar
augmented_image_path = r'D:\Document\Kuliah\Semster5\AI\TUBESAI\augmen_image'  # Path untuk menyimpan hasil augmentasi
categories = ['manusiareal', 'manusiaai', 'pemandanganreal', 'pemandanganai']  # Daftar kategori (folder) yang ada di dalam image_path
```

```python
# Membuat direktori untuk setiap kategori di folder augmented_image_path (jika belum ada)
for category in categories:
    os.makedirs(os.path.join(augmented_image_path, category), exist_ok=True)
```

```python
# Mendefinisikan augmentasi horizontal flip
horizontal_flip = A.HorizontalFlip(p=1.0)  # Selalu melakukan flip horizontal

# Fungsi untuk augmentasi dan menyimpan gambar
def augment_and_save_images():
    for category in categories:
        category_path = os.path.join(image_path, category)

        if not os.path.exists(category_path):
            print(f"Folder kategori {category} tidak ditemukan di path {category_path}")
            continue

        image_files = [f for f in os.listdir(category_path) if f.endswith('.jpg')]

        if len(image_files) == 0:
            print(f"Tidak ada file .jpg ditemukan di folder {category_path}")
            continue

        for file in image_files:
            file_path = os.path.join(category_path, file)

            # Memuat gambar
            try:
                image = cv2.imread(file_path)
                if image is None:
                    print(f"Error saat memuat file {file_path}: File tidak valid atau rusak.")
                    continue
            except Exception as e:
                print(f"Error saat memuat file {file_path}: {e}")
                continue

            # Augmentasi: Horizontal flip
            try:
                augmented = horizontal_flip(image=image)
                augmented_image = augmented['image']
            except Exception as e:
                print(f"Error saat melakukan augmentasi pada file {file_path}: {e}")
                continue

            # Pastikan folder tujuan ada
            augmented_category_path = os.path.join(augmented_image_path, category)
            os.makedirs(augmented_category_path, exist_ok=True)

            # Nama file untuk gambar yang telah di-augmentasi
            augmented_file_name = f'aug_flip_{file}'
            augmented_file_path = os.path.join(augmented_category_path, augmented_file_name)

            # Menyimpan gambar yang telah di-augmentasi
            try:
                # Memastikan bahwa file dapat disimpan dengan benar
                success = cv2.imwrite(augmented_file_path, augmented_image)
                if not success:
                    print(f"Warning: Gagal menyimpan file {augmented_file_path}")
            except Exception as e:
                print(f"Error saat menyimpan file {augmented_file_path}: {e}")
                continue

    print("Augmentasi selesai dan file gambar disimpan!")

# Menjalankan fungsi untuk augmentasi
augment_and_save_images()
```
