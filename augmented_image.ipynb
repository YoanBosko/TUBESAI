```python
import os
import cv2
import albumentations as A
```

```python
# Path dataset
input_folder = r"D:\Document\Kuliah\Semster5\AI\TUBESAI\dataset_image"
output_folder = r"D:\Document\Kuliah\Semster5\AI\TUBESAI\augmented_image"
categories = ['ManusiaAI', 'ManusiaReal', 'PemandanganAI', 'PemandanganReal']  # Daftar kategori (folder) yang ada di dalam image_path
```

```python
for category in categories:
    os.makedirs(os.path.join(output_folder, category), exist_ok=True)
```

```python
# Mendefinisikan augmentasi flip horizontal
horizontal_flip = A.HorizontalFlip(p=1.0)  # Augmentasi flip horizontal
```

```python
# Fungsi untuk augmentasi flip horizontal dan menyimpan gambar
def augment_and_save_horizontal_flip():
    for category in categories:
        category_path = os.path.join(input_folder, category)

        if not os.path.exists(category_path):
            print(f"Folder kategori {category} tidak ditemukan di path {category_path}")
            continue

        # Filter file dengan berbagai ekstensi gambar (case-insensitive)
        image_files = [f for f in os.listdir(category_path) if f.lower().endswith(('.jpg', '.jpeg', '.png', '.JPG'))]

        if len(image_files) == 0:
            print(f"Tidak ada file gambar ditemukan di folder {category_path}")
            continue

        for file in image_files:
            file_path = os.path.join(category_path, file)

            # Memuat gambar
            try:
                image = cv2.imread(file_path)
                image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Mengubah dari BGR ke RGB
            except Exception as e:
                print(f"Error saat memuat file {file_path}: {e}")
                continue

            # Augmentasi: Flip horizontal
            try:
                augmented_image = horizontal_flip(image=image)["image"]
            except Exception as e:
                print(f"Error saat melakukan augmentasi flip horizontal pada file {file_path}: {e}")
                continue

            # Menyimpan file hasil augmentasi flip horizontal
            augmented_category_path = os.path.join(output_folder, category)
            os.makedirs(augmented_category_path, exist_ok=True)
            augmented_file_name = f'aug_HorFlip_{file}'
            augmented_file_path = os.path.join(augmented_category_path, augmented_file_name)
            cv2.imwrite(augmented_file_path, cv2.cvtColor(augmented_image, cv2.COLOR_RGB2BGR))  # Simpan dalam format BGR

    print("Augmentasi flip horizontal selesai dan file gambar disimpan!")

# Menjalankan fungsi untuk augmentasi flip horizontal
augment_and_save_horizontal_flip()
```
